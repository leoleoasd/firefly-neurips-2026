```python
import json
results = json.load(open("../results/batch_summary.json"))
```

```python

# read task config 
task_configs = {
    k: json.load(open(f"../thirdparty/webarena/config_files/{k}.json"))
    for k in results['results'].keys()
}
```

```python

for key in results['results'].keys():
    results["results"][key]["intent"] = task_configs[key]["intent"]
    results["results"][key]["eval"] = task_configs[key]["eval"]
```

```python
json.dump(results, open("../results/batch_summary_with_config.json", "w"), indent=4)
```

```python

for key in results['results'].keys():
    if results["results"][key]["success"] == False and "string_match" not in results["results"][key]["eval"]["eval_types"]:
        print(key)
        print(json.dumps(results["results"][key], indent=4))
        print("-"*100)
```
